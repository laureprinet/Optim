**___Réponse Q3.___** 

Soit $\alpha >> 1$

Soit $i \in \{1, 2, \dots, p\}$

On a $h_i(q,d)=\frac{q_ie^{-\alpha q_i}+d_ie^{-\alpha d_i}}{e^{-\alpha q_i}+e^{-\alpha d_i}}$

Supposons $d_i \geq q_i$, $i.e. \min \{q_i,d_i\}=q_i$

On a alors :

$h_i(q,d)=q_i\frac{1+\frac{d_i}{q_i}e^{-\alpha (d_i-q_i)}}{e^{-\alpha (d_i-q_i)}}$

Or, $-\alpha (di-qi) << -1$ et $\frac{d_i}{q_i}<1$, donc $\frac{d_i}{q_i}e^{-\alpha (d_i-q_i)}<<1$

Ainsi, on a finalement $h_i(q,d) \approx q_i=\min \{q_i,d_i\}$

La fonction $h$ est donc une bonne approximation du minimum terme à terme.

L'intérêt de considérer ce problème approché plutôt que l'original est que $h$ est différentiable par rapport à $q$ et $d$.
