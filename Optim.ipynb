On s'intéresse à l'optimisation du profit d'un boulanger. Il a le contrôle sur la quantité de produit $q$ qu'il fabrique et $r$, la quantité en matière première qu'il commande !

**Formules importantes pour la suite :**

Nos variables : 

$q = (q_j) \in \mathbb{R}_+^p$

$v = (v_j) \in \mathbb{R}_+^p$

$d = (d_j) \in \mathbb{R}_+^p$

$r = (r_i) \in \mathbb{R}_+^m$

$c = (c_i) \in \mathbb{R}_+^m$

$A = (a_{i,j}) \in \mathcal{M}_{m,p}(\mathbb{R})$

Nos équations :

$$
r \geq Aq \tag{1} 
$$

$$
v^T \min \{q,d\}-c^Tr \tag{2}
$$

**___Réponse Q3.___** 

Soit $\alpha >> 1$

Soit $i \in \{1, 2, \dots, p\}$

On a $h_i(q,d)=\frac{q_ie^{-\alpha q_i}+d_ie^{-\alpha d_i}}{e^{-\alpha q_i}+e^{-\alpha d_i}}$

Sans perte de généralité, supposons $d_i \geq q_i$, $i.e. \min \{q_i,d_i\}=q_i$

On a alors :

$h_i(q,d)=q_i\frac{1+\frac{d_i}{q_i}e^{-\alpha (d_i-q_i)}}{e^{-\alpha (d_i-q_i)}}$

Or, $-\alpha (di-qi) << -1$ et $\frac{d_i}{q_i}<1$, donc $\frac{d_i}{q_i}e^{-\alpha (d_i-q_i)}<<1$

Ainsi, on a finalement $h_i(q,d) \approx q_i=\min \{q_i,d_i\}$

Puisque $d_i$ et $q_i$ ont des rôles symétriques, on a le résultat similaire si $q_i \geq d_i$

La fonction $h$ est donc une bonne approximation du minimum terme à terme.

L'intérêt de considérer ce problème approché plutôt que l'original est que $h$ est différentiable par rapport à $q$ et $d$.

```python
# Requirements 

import numpy as np
import casadi as cs
from casadi import Opti


# Paramètres d'entrée

alpha = 0.1
c = 10 ** (-3) * np.array([30, 1, 1.3, 4, 1])
v = np.array([0.9, 1.5, 1.1])
d = np.array([400, 67, 33])
A = np.array([[3.5, 2, 1], [250, 80, 25], [0, 8, 3], [0, 40, 10], [0, 8.5, 0]])

K = 3
pi = [0.5, 0.3, 0.2]
d_bis = [np.array([400, 67, 33]), np.array([500, 80, 53]), np.array([300, 60, 43])]
c_bis = [
    10 ** (-3) * np.array([30, 1, 1.3, 4, 1]),
    10 ** (-3) * np.array([40, 1.2, 1.2, 3, 1.2]),
    10 ** (-3) * np.array([20, 0.8, 1.4, 5, 0.8]),
]
```

**___Réponse Q6.___**

```python
#Q6

# Définition de la fonction h
def fun_h(a, q, d):
    h = (q * cs.exp(-a * q) + d * cs.exp(-a * d)) / (cs.exp(-a * q) + cs.exp(-a * d))
    return h

# Définition des variables pour minimisation
opti_cons = Opti()
q = opti_cons.variable(3)
r = opti_cons.variable(5)

# Minimisation de -f(q,r)
opti_cons.minimize(- cs.dot(v, fun_h(alpha, q, d)) + cs.dot(c, r))

# Contraintes à respecter
opti_cons.subject_to(-r <= 0)
opti_cons.subject_to(-q <= 0)
opti_cons.subject_to(A @ q - r <= 0)

opti_cons.solver("ipopt")
print("Solve Chocolatine with constraints with Casadi...")
sol = opti_cons.solve()
print(f"q = {sol.value(q)}")
print(f"r = {sol.value(r)}")
```

**___Réponse Q8.b___**

Supposons la contrainte $c_1$ inactive à l'optimum, avec $c_1(q,r)=Aq-r$
Ainsi, on a $Aq^{*}-r^{*}<0$ avec $(q^{*},r^{*})$ solution du problème formulé en **___Q8.a___**.

D'après les conditions de KKT pour un problème d'optimisation avec des contraintes égalité et inégalité, on a :

$$
\exists \lambda \in \mathbb{R}^{3} \text{ tel que } \nabla f(q^*,r^*) + \sum_{i=1}^{3} \lambda_i \nabla c_i(q^*,r^*) = 0 \tag{4}
$$
$$
\forall i \in \{1,2,3\}, c_i(q^*,r^*)\lambda_i=0 \text{ et } \lambda_i \geq 0, c_i(q^*,r^*) \leq 0 \tag{5}
$$

$$
\text{Où } c(q,r) = \begin{pmatrix}
Aq-r \\
-q \\
-r
\end{pmatrix}
\lambda_i \geq 0\
\text{ donc } \nabla c(q,r)=\begin{pmatrix}
A & -I & 0 \\
-I & 0 & -I 
\end{pmatrix}
\text{(matrice de taille (m+p,3))}
$$ 

De plus, on a :

$$
\nabla f(q,r)=\begin{pmatrix}
-v^T \frac{\partial h}{\partial q_1}(q,d)\\
...\\
-v^T \frac{\partial h}{\partial q_p}(q,d)\\
\\
\sum_{k=1}^{K}\pi_k c_1^k\\
...\\
\sum_{k=1}^{K}\pi_k c_m^k 
\end{pmatrix}
$$

Ainsi, $(4)$ devient :
$$
\forall j\in \llbracket 1, p \rrbracket, v^T \frac{\partial h}{\partial q_j}+\lambda_1A-\lambda_2I=0 \tag{6}
$$
$$
\forall i\in \llbracket 1, m \rrbracket, \sum_{k=1}^{K}\pi_k c_i^k -\lambda_1 I-\lambda_3 I =0 \tag{7}
$$

On a $\lambda_1=0$ par hypothèse donc $(6)$ devient : 

$$
\forall i\in \llbracket 1, m \rrbracket, \lambda_3I=-\sum_{k=1}^{K}\pi_k c_i^k
$$
Or, les $\{\pi_k\}$ sont $>0$ (si l'un d'eux est nul, le scénario n'est plus à considérer et on peut alors l'enlever de la somme). On a alors $\lambda_3 < 0$ ce qui contredit les conditions de KKT. On n'est donc pas à l'optimum.

Ainsi, on a nécessairement $c_1$ active à l'optimum.

$$\min_{q \in \mathbb{R^p}} f(q)$$

**___Réponse Q9.a___**

Montrons que résoudre le problème $(*)$ :
$$\max_{\left\{
\begin{array}{rl}
    (q,u_k) \in \mathbb{R}^p \times \mathbb{R}^p \\
  u_k - d_k \leq 0 \\

Or, $q^*$ étant donné, et puisque $v \geq 0$, $u_k = \min\{q^*,d_k\}$ est nécessairement optimal pour le problème $(*)$, on cherche donc bien simplement à trouver le $q^*$ maximisant $v^T min\{q,d_k\}$.

**___Réponse Q9.c___**


```python
# Définition des variables pour minimisation
opti_cons = Opti()
q = opti_cons.variable(3)
u0 = opti_cons.variable(3)
u1 = opti_cons.variable(3)
u2 = opti_cons.variable(3)
u = [u0, u1, u2]

# Minimisation de -f(q,r)
f = 0
for i in range(3):
    f += pi[i] * (cs.dot(v, u[i]) - cs.dot(c_bis[i], A @ q))
opti_cons.minimize(-f)

# Contraintes à respecter
opti_cons.subject_to(u[0] - q <= 0)
opti_cons.subject_to(u[1] - q <= 0)
opti_cons.subject_to(u[2] - q <= 0)
opti_cons.subject_to(u[0] - d <= 0)
opti_cons.subject_to(u[1] - d <= 0)
opti_cons.subject_to(u[2] - d <= 0)
opti_cons.subject_to(-q <= 0)

opti_cons.solver("ipopt")
print("Solve Chocolatine with constraints with Casadi...")
sol = opti_cons.solve()
print(f"q = {sol.value(q)}")
print(f"uk = {[sol.value(u0), sol.value(u1), sol.value(u2)]}")
```

**___Réponse Q10.a___**



Puisque la fonction min n'est pas différentiable, on choisit un algorithme de point fixe de l'opérateur proximal, qui est bien une méthode de résolution non-lisse.

*__Réponse Q10.b__*

```python
# Paramètres d'entrée
def fun(q):
    f = 0
    for k in range(3):
        f += (cs.dot(v, fun_h(alpha, q, d)) - cs.dot(c_bis[k], A @ q)) * pi[k]
    return f


# Définition de la fonction Prox
def prox(m, f, x):
    opti_cons = Opti()
    s = opti_cons.variable(len(x))
    opti_cons.minimize(f(s) + cs.norm_2(s - x) ** 2 / (2 * m))
    opti_cons.subject_to(opti_cons.bounded(0, s, 1000))
    opti_cons.solver("ipopt")
    sol = opti_cons.solve()
    return sol.value(s)


# Algorithme
def algo_prox(fun, epsilon):        # epsilon l'écart maximal entre x* et Prox(x*)
    q = np.array([1, 1, 1])
    l = 1                           # initialisation des l_k qui véirifient les conditions Ci) et Cii)
    c = True
    while c:
        y = prox(1 / l, fun, q)
        if cs.norm_2(y - q) < epsilon:  # lorsqu'on a trouvé un point fixe avec une précision suffisante, on sort de la boucle
            c = False
        else:
            l += 1
            q = y
    return q
```
